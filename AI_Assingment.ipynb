```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load cleaned data
df = pd.read_csv('card_transdata_cleaned.csv')

# Plot Class (Fraud) Distribution
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='fraud', palette='Set2')
plt.title('Distribution of Fraud Cases (0 = Legitimate, 1 = Fraud)')
plt.xlabel('Fraud Status')
plt.ylabel('Count')
plt.xticks([0, 1], ['Legitimate (0)', 'Fraud (1)'])
plt.tight_layout()
plt.show()
```

```python
# Plot Used Pin Number Distribution
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='used_pin_number', palette='coolwarm')
plt.title('Distribution of PIN Usage')
plt.xlabel('Used PIN Number')
plt.ylabel('Count')
plt.xticks([0, 1], ['No PIN (0)', 'Used PIN (1)'])
plt.tight_layout()
plt.show()
```

```python
# Plot Repeat Retailer Distribution
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='repeat_retailer', palette='pastel')
plt.title('Distribution of Repeat Retailer Transactions')
plt.xlabel('Repeat Retailer')
plt.ylabel('Count')
plt.xticks([0, 1], ['New Retailer (0)', 'Repeat Retailer (1)'])
plt.tight_layout()
plt.show()
```

```python
# Plot Online Order Distribution
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='online_order', palette='viridis')
plt.title('Distribution of Online vs. In-Person Orders')
plt.xlabel('Online Order')
plt.ylabel('Count')
plt.xticks([0, 1], ['In-Person (0)', 'Online (1)'])
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 4))
sns.histplot(df['ratio_to_median_purchase_price'], bins=30, kde=True, color='purple')
plt.title('Distribution of Ratio to Median Purchase Price')
plt.xlabel('Ratio to Median Purchase Price')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='distance_from_home', y='ratio_to_median_purchase_price', hue='fraud', palette={0: 'blue', 1: 'orange'}, alpha=0.7)
plt.title('Bivariate 1: Distance from Home vs. Ratio to Median Price')
plt.xlabel('Distance from Home')
plt.ylabel('Ratio to Median Purchase Price')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='distance_from_last_transaction', y='distance_from_home', hue='fraud', palette={0: 'blue', 1: 'orange'}, alpha=0.7)
plt.title('Bivariate 2: Distance from Last Transaction vs. Distance from Home')
plt.xlabel('Distance from Last Transaction')
plt.ylabel('Distance from Home')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='distance_from_last_transaction', y='ratio_to_median_purchase_price', hue='fraud', palette={0: 'blue', 1: 'orange'}, alpha=0.7)
plt.title('Bivariate 3: Distance from Last Transaction vs. Ratio to Median Price')
plt.xlabel('Distance from Last Transaction')
plt.ylabel('Ratio to Median Purchase Price')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='fraud', y='distance_from_home', palette='Set2')
plt.title('Bivariate 4: Distance from Home by Fraud Status')
plt.xlabel('Fraud Status (0 = Legitimate, 1 = Fraud)')
plt.ylabel('Distance from Home')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='fraud', y='ratio_to_median_purchase_price', palette='Set3')
plt.title('Bivariate 5: Ratio to Median Price by Fraud Status')
plt.xlabel('Fraud Status (0 = Legitimate, 1 = Fraud)')
plt.ylabel('Ratio to Median Purchase Price')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x='online_order', y='fraud', errorbar=None, palette='coolwarm')
plt.title('Bivariate 6: Fraud Rate by Online Order Status')
plt.xlabel('Online Order (0 = In-Person, 1 = Online)')
plt.ylabel('Fraud Rate')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df, 
    x='distance_from_home', 
    y='ratio_to_median_purchase_price', 
    hue='fraud', 
    palette={0: 'green', 1: 'red'}, 
    alpha=0.8,
    s=60
)

plt.title('Fraudulent vs. Non-Fraudulent Purchases', fontsize=14)
plt.xlabel('Distance from Home', fontsize=12)
plt.ylabel('Ratio to Median Purchase Price', fontsize=12)

# Custom legend / key
handles, labels = plt.gca().get_legend_handles_labels()
plt.legend(handles=handles, labels=['Non-Fraud (0)', 'Fraud (1)'], title='Transaction Type', title_fontsize='11', fontsize='10')

plt.tight_layout()
plt.show()
```
